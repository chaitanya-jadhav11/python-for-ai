```python
# difference_update()
# Action: Removes elements from the original set that are also present in the specified other set(s).
```

```python
A = {1, 2, 3, 4}
B = {3, 4, 5, 6}

B.difference_update(A)
# or A -= B 

print(B) # Output: {1, 2} (A is modified)
```
