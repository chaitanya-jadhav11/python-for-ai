```python
class Dog:
    
    # Class Object Attribute its like Static in Java/ value same for all object
    species = 'mammal'
    
    def __init__(self,breed,name):
        self.breed = breed
        self.name = name

    def bark(self, number):
        print ('bark. my name is {}'.format(self.name))
        print ('bark. my species is {}'.format(self.species))
        print ('bark. my number is {}'.format(number))
```

```python
myDog = Dog(breed='Lab', name= 'xyz')
type(myDog)
```

```python
myDog.breed
myDog.name
```

```python
myDog.species
```

```python
myDog.bark(22)
```

```python
class Circle():
    pi = 3.14

    def __init__(self,radius=1):
        self.radius= radius;
        self.area= radius * radius * self.pi
        # self.area= radius * radius * Circle.pi Same as above
    def get_circ(self):
        return self.radius * self.pi * 2
```

```python
myCircle = Circle(30)
```

```python
myCircle.pi
```

```python
myCircle.get_circ()
```

```python
myCircle.area
```
