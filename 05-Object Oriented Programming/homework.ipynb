Fill in the Line class methods to accept coordinates as a pair of tuples and return the slope and distance of the line.

```python
import math

def calculate_distance(p1, p2):
    """
    Calculates the Euclidean distance between two points (p1, p2) 
    where each point is an iterable of coordinates (e.g., a tuple or list).
    """
    # math.hypot handles n-dimensions and is efficient
    return math.hypot(p2[0] - p1[0], p2[1] - p1[1]) 

def slope(x1, y1, x2, y2):
    return (y2-y1)/(x2-x1)
```

```python
class Line:
    
    def __init__(self,coor1,coor2):
        self.coor1= coor1
        self.coor2= coor2
    
    def distance(self):
        return calculate_distance(self.coor1, self.coor2)
    
    def slope(self):
        return slope(self.coor1[0],self.coor1[1],self.coor2[0],self.coor2[1])
```

```python
# EXAMPLE OUTPUT

coordinate1 = (3,2)
coordinate2 = (8,10)

li = Line(coordinate1,coordinate2)
```

```python
li.distance()
```

```python
li.slope()
```
