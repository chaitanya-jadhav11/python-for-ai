```python
# Generator function for the cube of numbers (power of 3)
def creare_cubes(n):
    result =[]
    for num in range(n):
        result.append( num**3)
    return result   
```

```python
print (creare_cubes(10)) # create full List
```

Generators

```python
# Generator function for the cube of numbers (power of 3)
def gencubes(n):

    for num in range(n):
        yield num**3
```

```python
for x in gencubes(10):
    print(x)
```

```python
def simple_gen():
    for x in range (3):
        yield x
```

```python
for number in simple_gen():
    print (number)
```
