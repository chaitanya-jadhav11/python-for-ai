Problem 1

Create a generator that generates the squares of numbers up to some number N.

```python
def gensquares(n):
    for n in range(n):
        yield n * n
```

```python
sq_var= gensquares(3)
```

```python
next(sq_var)
```

```python
next(sq_var)
```

Problem 2

Create a generator that yields "n" random numbers between a low and high number (that are inputs).
Note: Use the random library. For example:

```python
import random

random.randint(1,10)

def rand_num(low,high,n):
    for n in range(n):
        yield random.randint(low,high)
    
    
```

```python
for num in rand_num(1,10,12):
    print(num)
```

```python
my_list = [1,2,3,4,5]

gencomp1 = (item for item in my_list if item > 3) #  Generator expressions make the most sense in scenarios where you need to take one item at a time, 

for item in gencomp1:
    print(item)
```
