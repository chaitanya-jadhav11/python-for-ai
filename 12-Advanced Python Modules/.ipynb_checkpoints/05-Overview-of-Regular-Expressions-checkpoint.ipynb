```python
text = "The person's phone number is 408-555-1234. Call soon!"
```

```python
# other way to find str 
# 1
'phone' in text

import re
```

```python
pattern = 'Not in text'
re.search(pattern,text) # no return 
```

```python
pattern = 'phone'
re.search(pattern,text)
```

```python
match = re.search(pattern,text)
```

```python
match.span()
```

```python
match.start()
```

```python
text = "my phone is a new phone"
match = re.search("phone",text)
match.span() #Notice it only matches the first instance. If we wanted a list of all matches, we can use .findall() method:
```

```python
matches = re.findall("phone",text)
len(matches)
for match in re.finditer("phone",text):
    print(match.span())
```

### Patterns

```python
text = "My telephone number is 408-555-1234"
phone = re.search(r'\d\d\d-\d\d\d-\d\d\d\d',text)
```

```python
phone
```

```python
phone.group()
```

```python
text1 = "My telephone number is 408-555-1234"
phone1 = re.search(r'\d{3}-\d{3}-\d{4}',text)
phone1
```

### Additional Regex Syntax

```python
re.search(r"man|woman","This man was here.")   # OR 
```

```python
re.search(r"man|woman","This woman was here.")
```

The Wildcard Character

```python
re.findall(r".at","The cat in the hat sat here.")
```

```python
re.findall(r"...at","The bat went splat")
```

Starts with and Ends With

```python
# Ends with a number
re.findall(r'\d$','This ends with a number 2')
```

```python
# Starts with a number
re.findall(r'^\d','1 is the loneliest number.')
```

Exclusion

```python
phrase = "there are 3 numbers 34 inside 5 this sentence."
```

```python
re.findall(r'[^\d]',phrase)
```
