```python
from collections import Counter
```

```python
lst = [1,2,2,2,2,3,3,3,1,2,1,12,3,2,32,1,21,1,223,1]

Counter(lst) # count each item occurances
```

```python
Counter('aabsbsbsbhshhbbsbs')
```

```python
strList = ['a','a','b','c','c']
```

```python
Counter(strList)
```

```python
s = 'How many times does each word show up in this sentence word times each each word'
```

```python
Counter(s)
```

```python
words = s.split()

Counter(words)
```

<b> Common method when using the Counter() object </b>

```python
counter_obj=Counter(strList)
```

```python
counter_obj.most_common()
counter_obj.most_common(1)
```

```python
sum(counter_obj.values())
```

```python
list(counter_obj)
```

<b>defaultdict</b>

```python
from collections import defaultdict
```

```python
d = defaultdict(lambda: 0) # Set default value to dict
```

```python
d['d']='d'
```

```python
d['d']
```

```python
d['d1']
```

<b >namedtuple </b>

```python
from collections import namedtuple
```

```python
Dog = namedtuple('Dog',['age','breed','name'])

sam = Dog(age=2,breed='Lab',name='Sammy')

frank = Dog(age=2,breed='Shepard',name="Frankie")
```
