```python
try:
    result = 10 + 10
except:
    print ('Error')
else :
    print ('result {}'.format(result))
    
print('scripts end here')
```

```python
try :
    f = open('testfile','r')
    f.write('hi there')

except TypeError:
    print (' Type Error')
except OSError:
    print ('UnsupportedOperation error')
except:     
    print ('catch generic error')
finally:
    print ('in finally block for cleaning reources')

    
print('scripts end here')   
```
